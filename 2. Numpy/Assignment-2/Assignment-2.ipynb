## Exercise 1 — Array creation
Create a NumPy array containing the numbers 5, 10, 15, 20, 25. Print its:

## Exercise 2 — Indexing and slicing
Given:
```python
values = np.array([12, 24, 36, 48, 60, 72, 84, 96])
```
1. Print the first element.
2. Print the last element.
3. Print elements from index 2 to index 5 (excluding 5).
4. Print every second element.
5. Print the array in reverse order.


````python
values = np.array([12, 24, 36, 48, 60, 72, 84, 96])
print("Original values:", values)

# 1. Print the first element
print("1. First element:", values[0])

# 2. Print the last element
print("2. Last element:", values[-1])

# 3. Print elements from index 2 to index 5 (excluding 5)
print("3. Elements from index 2 to 4:", values[2:5])

# 4. Print every second element
print("4. Every second element:", values[::2])

# 5. Print the array in reverse order
print("5. Reversed array:", values[::-1])
````

## Exercise 3 — 2D array
Create this array:
```text
10 20 30
40 50 60
70 80 90
```
Then:
1. Print the value 50.
2. Print the second row.
3. Print the third column.
4. Calculate the sum of each row.
5. Calculate the average of each column.


````python
# Create 2D array
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("2D Array:")
print(matrix)

# 1. Print the value 50 (row index 1, column index 1)
print("\n1. Value 50:", matrix[1, 1])

# 2. Print the second row (row index 1)
print("2. Second row:", matrix[1])

# 3. Print the third column (all rows, column index 2)
print("3. Third column:", matrix[:, 2])

# 4. Calculate the sum of each row (axis=1)
print("4. Sum of each row:", matrix.sum(axis=1))

# 5. Calculate the average of each column (axis=0)
print("5. Average of each column:", matrix.mean(axis=0))
````

## Exercise 4 — Filtering
Given:
```python
ages = np.array([15, 22, 17, 30, 16, 25, 19])
```
1. Select ages greater than or equal to 18.
2. Select ages below 18.
3. Select ages between 18 and 25 inclusive.


````python
ages = np.array([15, 22, 17, 30, 16, 25, 19])
print("Ages:", ages)

# 1. Select ages greater than or equal to 18
adults = ages[ages >= 18]
print("1. Ages >= 18:", adults)

# 2. Select ages below 18
minors = ages[ages < 18]
print("2. Ages < 18:", minors)

# 3. Select ages between 18 and 25 inclusive
young_adults = ages[(ages >= 18) & (ages <= 25)]
print("3. Ages between 18 and 25 (inclusive):", young_adults)
````

## Exercise 5 — Reshaping
Create numbers from 1 to 20. Reshape them into 4 rows and 5 columns. Then print the first two rows.


````python
# 1. Create numbers from 1 to 20
numbers = np.arange(1, 21)
print("Original 1D array (1 to 20):")
print(numbers)

# 2. Reshape into 4 rows and 5 columns
grid = numbers.reshape(4, 5)
print("\nReshaped 4x5 array:")
print(grid)

# 3. Print the first two rows
print("\nFirst two rows:")
print(grid[:2])
````

## Exercise 6 — Sales analysis
A shop records daily sales:
```python
sales = np.array([1200, 1500, 900, 1800, 2000, 1100, 1700])
```
Find:
1. Total sales.
2. Average sales.
3. Highest and lowest sales.
4. Days with sales greater than 1500.
5. Sales after a 10% increase.


````python
sales = np.array([1200, 1500, 900, 1800, 2000, 1100, 1700])
print("Daily Sales:", sales)

# 1. Total sales
total_sales = sales.sum()
print("1. Total sales:", total_sales)

# 2. Average sales
avg_sales = sales.mean()
print("2. Average sales:", round(avg_sales, 2))

# 3. Highest and lowest sales
print("3. Highest sales:", sales.max(), "| Lowest sales:", sales.min())

# 4. Days with sales greater than 1500
high_sales = sales[sales > 1500]
print("4. Sales greater than 1500:", high_sales)

# Day numbers where sales were > 1500 (1-based day index)
day_numbers = np.where(sales > 1500)[0] + 1
print("   Recorded on Days:", day_numbers)

# 5. Sales after a 10% increase
increased_sales = sales * 1.10
print("5. Sales after 10% increase:", increased_sales)
````

## Exercise 7 — Student result system
Create a 2D array for marks of four students in three subjects. Calculate:
1. Total marks per student.
2. Average marks per student.
3. Highest mark in each subject.
4. Which students scored at least 50 in every subject.


````python
# 4 students (rows), 3 subjects (columns)
# Rows: Student 1, Student 2, Student 3, Student 4
# Columns: Subject 1, Subject 2, Subject 3
marks = np.array([
    [75, 80, 85],   # Student 1
    [45, 60, 55],   # Student 2 (Scored 45 in Subject 1)
    [90, 88, 92],   # Student 3
    [52, 50, 70]    # Student 4
])

print("Student Marks (Rows = Students, Columns = Subjects):")
print(marks)

# 1. Total marks per student (sum along axis=1)
total_per_student = marks.sum(axis=1)
print("\n1. Total marks per student:", total_per_student)

# 2. Average marks per student (mean along axis=1)
avg_per_student = marks.mean(axis=1)
print("2. Average marks per student:", np.round(avg_per_student, 2))

# 3. Highest mark in each subject (max along axis=0)
highest_per_subject = marks.max(axis=0)
print("3. Highest mark in each subject:", highest_per_subject)

# 4. Which students scored at least 50 in every subject
passed_all = np.all(marks >= 50, axis=1)
print("4. Passed all subjects (marks >= 50):", passed_all)

# Display student numbers (1-indexed)
student_ids = np.where(passed_all)[0] + 1
print("   Students who scored at least 50 in all subjects: Students", list(student_ids))
````

## Exercise 8 — Challenge: Temperature analysis
Given a 2D array where each row represents one week and each column represents one day, calculate:
1. Average temperature per week.
2. Average temperature per day across all weeks.
3. The highest temperature in the dataset.
4. All temperatures above 35.


````python
# 3 weeks (rows), 7 days (columns)
temperatures = np.array([
    [32, 34, 36, 33, 31, 35, 37],  # Week 1
    [29, 30, 32, 35, 38, 36, 34],  # Week 2
    [33, 35, 37, 39, 40, 36, 32]   # Week 3
])

print("Temperatures (3 Weeks x 7 Days):")
print(temperatures)

# 1. Average temperature per week (mean along axis=1)
avg_per_week = temperatures.mean(axis=1)
print("\n1. Average temperature per week:", np.round(avg_per_week, 2))

# 2. Average temperature per day across all weeks (mean along axis=0)
avg_per_day = temperatures.mean(axis=0)
print("2. Average temperature per day (across 3 weeks):", np.round(avg_per_day, 2))

# 3. The highest temperature in the dataset
max_temp = temperatures.max()
print("3. Highest temperature:", max_temp)

# 4. All temperatures above 35
hot_temps = temperatures[temperatures > 35]
print("4. Temperatures above 35:", hot_temps)
````
