```python
class BankAccount:
    def __init__(self, name: str, account_number: str, balance: int = 0):
        if not name.strip():
            raise ValueError("Name cannot be empty")
        if isinstance(balance, bool) or not isinstance(balance, int) or balance < 0:
            raise ValueError("Balance must be a non-negative integer")

        self._name = name                 # Convention: intended for internal use
        self.__account_number = account_number
        self.__balance = balance

    @property
    def balance(self) -> int:
        return self.__balance

    @property
    def masked_account_number(self) -> str:
        return "****" + self.__account_number[-4:]

    def deposit(self, amount: int) -> None:
        self.__validate_amount(amount)
        self.__balance += amount

    def withdraw(self, amount: int) -> None:
        self.__validate_amount(amount)
        if amount > self.__balance:
            raise ValueError("Insufficient funds")
        self.__balance -= amount

    def __validate_amount(self, amount: int) -> None:
        if isinstance(amount, bool) or not isinstance(amount, int) or amount <= 0:
            raise ValueError("Amount must be a positive integer")


user_account = BankAccount("Ali", "1010", 1000)
user_account.deposit(200)
print(user_account.balance)                 # 1200
print(user_account.masked_account_number)   # ****1010
```

```python
def custom_print(*args, separator=", "):
    print(separator.join(str(arg) for arg in args))

custom_print(1, 2, 3)
custom_print("apple", "banana", "cherry", separator=" | ")
```

```python
def average_grades(*grades):
    if grades:
        return sum(grades) / len(grades)
    return 0

print(average_grades(90, 80, 85, 70))
print(average_grades())
```

```python
def add(data_type, *args):
    if data_type == 'int':
        return sum(args) 
    elif data_type == 'str':
        return "".join(args)
    else:
        raise ValueError("Unsupported data type. Use 'int' or 'str'.")

print(add('int', 5, 7))     
print(add('str', 'Hi ', 'Geeks')) 
```
