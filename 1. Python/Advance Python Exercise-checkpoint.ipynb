## Find the plaindrome
## A string that is same in both directions
## EG: madam is a plaindrome

```python
def checkPlaindrome(strr):
    originalSr = []
    plaindromeCandidate = []
    indddexx = 0
    for char in strr:
        originalSr.append(char)
        
    # reverse String
    reversedIndex = len(originalSr)
    reversed
    for i in range(len(originalSr)):
        indexx = reversedIndex - i -1
        plaindromeCandidate.append(originalSr[indexx])
    
    # item by item comparision
    indddexx == 0
    for item in originalSr:
        if item == plaindromeCandidate[indddexx]:
            print(f"item: {item}  ===== {plaindromeCandidate[indddexx]}")
        else:
            print(f"item: {item}  ,. {plaindromeCandidate[indddexx]}")
        indddexx = indddexx + 1
    
    
    if originalSr == plaindromeCandidate:
        print("String is Plaindrome")
    else:
        print("String is not Plaindrome")        
        
    print("originalSr   :        : ", originalSr)
    print("plaindromeCandidate   : ", plaindromeCandidate)
```

```python
checkPlaindrome("mafgdfddam")
```

# Anagram

## two strings are same 

## listen = silent
## heart = earth
## evil = vile
## hello != world

```python
def anagram(str1, str2):
    A1 = []
    A2 = []
    
    countA1 = 0
    countA2 = 0
    
    foundCount = 0
    
    correctItemCount = 0
        
    for i in str1:
        countA1 += 1
        A1.append(i)
        
    for j in str2:
        countA2 += 1
        A2.append(j)
    
    # check count of both  arrays
    
    if countA1 == countA2:
        
        for i in A1:
            for j in A2:
                if i == j:
                    foundCount = foundCount + 1
                    break
    else:
        print("Not Anagram")
        
    if foundCount == countA1:
        print("Anagram")
    else:
        print("Not Anagram")
                    
anagram("silent", "listen")
anagram("heart", "earth")
anagram("heart", "ttttt")
```

# Contains Duplicate

## Check for array if it contains duplicate

```python
def duplicateCheck(arr):
    duplicatFound = False
    
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] == arr[j]:
                duplicatFound = True
                print("Duplicate = ", arr[i])
        
    if duplicatFound:
        print("Array has Duplicates")
    else:
        print("No duplicates")
        
duplicateCheck([1,2,3,4,5,3,7,8,9, 2])
```
