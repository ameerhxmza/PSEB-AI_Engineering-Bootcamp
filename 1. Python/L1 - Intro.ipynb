```python

# 1. Addition
def sum(a, b):
    return a + b


# 2. Subtraction
def sub(a, b):
    return a - b


# 3. Multiplication
def mul(a, b):
    return a * b


# 4. Division
def div(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b


# 5. Orchestrator Function
def calculator(a, b, operation):

    if operation == "sum":
        return sum(a, b)

    elif operation == "sub":
        return sub(a, b)

    elif operation == "mul":
        return mul(a, b)

    elif operation == "div":
        return div(a, b)

    else:
        return "Invalid operation"


# 6. Input Function
def get_input():

    num1 = input("Enter first number: ")
    num2 = input("Enter second number: ")

    return num1, num2


# 7. Type Casting Function
def type_casting():

    try:
        num1, num2 = get_input()

        num1 = float(num1)
        num2 = float(num2)

        operation = input(
            "Choose operation (sum, sub, mul, div): "
        ).lower()

        result = calculator(num1, num2, operation)

        print("Result:", result)

    except ValueError:
        print("Error: Invalid number! Restarting...\n")
        type_casting()


# 8. Main Function
def main():
    type_casting()


# Execute Program
main()
```

```python
numbers = [45, 12, 89, 3, 12, 67, 24, 9, 3, 56, 1, 34, 45]
```

```python
uN = []

for i in range(len(numbers)):
    
    duplicate = False
    
    for j in range(len(uN)):
        if numbers[i] == uN[j]:
            duplicate = True
            break
    
    if duplicate == False:
        uN.append(numbers[i])
            
print(uN)
        

```
