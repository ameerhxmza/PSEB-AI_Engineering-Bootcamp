# Variable Scope

```python
# Local Scope 
def anyFunc():
    var1 = "This is var 1"
    print(var1)
    print(var1)
anyFunc()
```

```python
# Global scope

var1 = 10
def anyFunc():
    global var1
    var1 = 20
    print(var1)
anyFunc()
print(var1)
```

```python
count = 0
def increment():
    global count
    count = count + 1
    
for i in range(20):
    increment()
print(count)
```

```python
# Enclosing scope

def outerFunction():
    count = 0
    def innerFunction():
        nonlocal count 
        count = count + 1
        
    innerFunction()
    print(count)
    
outerFunction()
```
